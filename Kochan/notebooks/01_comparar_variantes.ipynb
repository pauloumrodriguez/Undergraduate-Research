# 01 — Comparação das variantes de rede

Este notebook lê a tabela consolidada produzida pelo pipeline. A comparação separa aproximação numérica de fidelidade metodológica.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

RAIZ = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
comparacao = pd.read_csv(RAIZ / 'evidencias' / 'comparacao_variantes.csv')
colunas = ['variante', 'avaliacao', 'papel', 'erro_relativo_medio', 'provas_tcc_aprovadas', 'gate_modelagem']
comparacao[colunas].sort_values('erro_relativo_medio').head(15)
```

```python
grafico = (comparacao.dropna(subset=['erro_relativo_medio'])
          .sort_values('erro_relativo_medio')
          .drop_duplicates('variante'))
cores = ['#b23a48' if papel == 'ajuste_pos_hoc' else '#2a6f97' for papel in grafico['papel']]
ax = grafico.plot.barh(x='variante', y='erro_relativo_medio', color=cores, legend=False, figsize=(9, 6))
ax.set_xlabel('Erro relativo médio')
ax.set_ylabel('Variante')
ax.set_title('Proximidade das contagens não equivale a confirmação metodológica')
plt.tight_layout()
```
