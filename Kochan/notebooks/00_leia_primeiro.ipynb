Este notebook apresenta a replicação nova sem executar consultas pesadas no Solr. Ele mostra as entradas, as regras declaradas e os resultados já produzidos.

```python
from pathlib import Path
import json

RAIZ = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
print('Pasta da replicação:', RAIZ.resolve())
for nome in ['dados', 'referencias', 'config', 'src', 'notebooks', 'runs', 'evidencias', 'tests', 'docs']:
    caminho = RAIZ / nome
    print(f'{nome:12} ->', 'presente' if caminho.exists() else 'AUSENTE')
```
