# 02 — Inspeção da rede e preparação do classificador

A rede v11 é usada aqui porque foi a melhor candidata empírica não ajustada diretamente aos totais. Isso não significa que ela seja a rede original de Patrick.

```python
from pathlib import Path
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt

RAIZ = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
arquivo = RAIZ / 'runs' / 'redes' / 'v11_apenas_termo_principal' / 'relacoes_2006.csv'
arestas = pd.read_csv(arquivo)
arestas.head()
```

```python
G = nx.from_pandas_edgelist(arestas, 'origem_rotulo', 'destino_rotulo', edge_attr='coocorrencias')
print('Nós:', G.number_of_nodes())
print('Arestas:', G.number_of_edges())
print('Componentes:', nx.number_connected_components(G))
graus = sorted(G.degree, key=lambda item: item[1], reverse=True)
graus[:10]
```

```python
principais = [no for no, _ in graus[:20]]
H = G.subgraph(principais).copy()
plt.figure(figsize=(12, 9))
pos = nx.spring_layout(H, seed=42)
nx.draw_networkx(H, pos, node_size=900, font_size=8, edge_color='#9aa0a6', node_color='#8ecae6')
plt.title('Subgrafo dos 20 nós de maior grau — v11, 2006')
plt.axis('off')
plt.tight_layout()
```
